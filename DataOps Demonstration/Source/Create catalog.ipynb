```python
%sql
CREATE CATALOG IF NOT EXISTS dqx_dev;
CREATE CATALOG IF NOT EXISTS dqx_prod;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS dqx_dev.customer;
CREATE SCHEMA IF NOT EXISTS dqx_prod.customer;
```

```python
%sql
CREATE TABLE IF NOT EXISTS dqx_dev.customer.customer_raw
(
    customer_id BIGINT,
    customer_name STRING,
    email STRING,
    age INT,
    address STRING,
    country STRING,
    gender STRING,
    customer_type STRING,
    created_date DATE
)
USING DELTA;
```
